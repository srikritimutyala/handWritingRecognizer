# Generate dataset

```python
import cv2
import csv
import glob

header = ["label"]
for i in range(0,2500):
    header.append("pixel"+str(i))
with open('dataset.csv', 'a') as f:
    writer = csv.writer(f)
    writer.writerow(header)

for label in range(10):
    dirList = glob.glob("/Users/Srikriti/Documents/HandWrittingRecognition/captured_images/"+str(label)+"/*.png")
    for img_path in dirList:
        im = cv2.imread(img_path)
        im_gray = cv2.cvtColor(im,cv2.COLOR_BGR2GRAY)
        im_gray = cv2.GaussianBlur(im_gray,(15,15),0)
        roi = cv2.resize(im_gray, (50,50), interpolation = cv2.INTER_AREA)

        data = []
        data.append(label)
        rows, cols = roi.shape


        for i in range(rows):
            for j in range(cols):
                k = roi[i,j]
                if k>100:
                    k = 1
                else:
                    k = 0
                data.append(k)
        with open('dataset.csv', 'a') as f:
            writer = csv.writer(f)
            writer.writerow(data)
```

# Load dataset

```python
import pandas as pd
from sklearn.utils import shuffle
data = pd.read_csv("/Users/Srikriti/Documents/HandWrittingRecognition/dataset.csv")
data
```

# Seperation of dependant and independant variable

```python
X = data.drop(["label"], axis = 1)
Y = data["label"]
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import cv2
idx = 200
img = X.loc[idx].values.reshape(50,50)
print(Y[idx])
plt.imshow(img)
```

# train test split

```python
conda install sklearn.model_selection
```

```python
from sklearn.model_selection import train_test_split
train_x, test_x, train_y, test_y = train_test_split(X,Y, test_size = 0.2)
```

```python
import joblib
from sklearn.svm import SVC
classifier = SVC(kernel = "linear", random_state = 6)
classifier.fit(train_x, train_y)
joblib.dump(classifier,"/Users/Srikriti/Documents/HandWrittingRecognition/model/digit_recognizer")
```

# calculate accuracy

```python
from sklearn import metrics
prediction = classifier.predict(test_x)
print("Accuracy: ", metrics.accuracy_score(prediction,test_y))
```

# prediction of image drawn in paint

```python
conda install numpy
```

```python
import joblib
import cv2
import numpy as np
import time
import pyscreenshot as ImageGrab
model = joblib.load("/Users/Srikriti/Documents/HandWrittingRecognition/model/digit_recognizer")
image_folder = "/Users/Srikriti/Documents/HandWrittingRecognition/img/"


while True:
    img = ImageGrab.grab(bbox=(60,170,400,550))
    img.save(image_folder+"img.png")
    im = cv2.imread(image_folder+"img.png")
    im_gray = cv2.cvtColor(im, cv2.COLOR_BGR2GRAY)
    im_gray = cv2.GaussianBlur(im_gray,(15,15),0)

    #Threshold the image
    ret, im_th = cv2.threshold(im_gray, 100, 255, cv2.THRESH_BINARY)
    roi = cv2.resize(im_gray, (50,50), interpolation = cv2.INTER_AREA)
    X = []
    rows, cols = roi.shape


    for i in range(rows):
        for j in range(cols):
            k = roi[i,j]
            if k>100:
                k = 1
            else:
                k = 0
            X.append(k)


    predictions = model.predict([X])
    print("Prediction: ",predictions[0])
    cv2.putText(im, "Prediction: "+str(predictions[0]), (20,20), 0,0.8,(0,255,0),2,cv2.LINE_AA)
    cv2.startWindowThread()
    cv2.namedWindow("Result")
    cv2.imshow("Result",im)
    cv2.waitKey(10000)
    if cv2.waitKey(1) ==27:
        break
cv2.destroyAllWindows()
        
        
    
    

    

    
    
```
